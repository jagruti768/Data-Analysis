```python
import pandas as pd
df = pd.read_csv('data.csv')
print(df.head())
```

```python
# Display information about the dataset
print(df.info())

# Check the number of rows and columns
print(df.shape)

# Check column names
print(df.columns)

# Check missing values
print(df.isnull().sum())
```

```python
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

print(df["Date"].head())
```

```python
print(df.isnull().sum())
```

```python
df["Calories"] = df["Calories"].fillna(df["Calories"].mean())
```

```python
df = df.dropna(subset=["Date"])
```

```python
print(df.isnull().sum())
```

```python
print("Duplicate rows:", df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicate rows after cleaning:", df.duplicated().sum())
```

```python
duration_60 = df[df["Duration"] == 60]

print(duration_60)
```

```python
high_calories = df[df["Calories"] > 300]

print(high_calories)
```

```python
high_pulse = df[df["Pulse"] > 110]

print(high_pulse)
```

```python
result = df[(df["Duration"] == 60) & (df["Calories"] > 300)]

print(result)
```

```python
df["Calories_per_Minute"] = df["Calories"] / df["Duration"]

print(df.head())
```

```python
df["Pulse_Difference"] = df["Maxpulse"] - df["Pulse"]

print(df.head())
```

```python
df["Intensity"] = df["Pulse"].apply(
    lambda x: "High" if x >= 110 else "Normal"
)

print(df[["Pulse", "Intensity"]].head())
```

```python
print(df[df["Duration"] > 120])
```

```python
df = df[df["Duration"] <= 120]
```

```python
df.to_csv("cleaned_data.csv", index=False)

print("Cleaned dataset saved successfully.")
```
